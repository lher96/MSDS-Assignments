# Krackhardt Kite Graph using NetworkX

This notebook creates a basic graph using Python and the NetworkX package. The example is based on the Krackhardt Kite graph which appears to be a common basic graph.
 Video Link:https://www.youtube.com/watch?v=BQ95u6WMilA

##Setup



```python
import networkx as nx
import matplotlib.pyplot as plt
```

## Define nodes

```python
#(Andre), (Beverley), (Carol), (Diane), (Ed), (Fernando), (Garth), (Heather), (Ike), and (Jane) I have replaced with my own name
#initially I used person1-10 but it seems from several online examples these names are used similarly to your graph. I simply replaced the final name Jane
nodes = [
    "Andre", "Beverly", "Carol", "Diane", "Ed",
    "Fernando", "Garth", "Heather", "Ike", "Luis"
]

nodes
```

## Define edges

```python
#used tuples here instead of lists or dictionaries ideally for speed though not critical for this
#it makes sense since tuples are pairs and are immutable. For extremely large scale I believe a 2D numpy array would be fastest
#this is very similar to a leetcode problem where you must find a path to a node
edges = [
    ("Andre", "Beverly"),
    ("Andre", "Carol"),
    ("Andre", "Diane"),
    ("Andre", "Fernando"),
    ("Beverly", "Diane"),
    ("Beverly", "Ed"),
    ("Beverly", "Garth"),
    ("Carol", "Diane"),
    ("Carol", "Fernando"),
    ("Diane", "Ed"),
    ("Diane", "Fernando"),
    ("Diane", "Garth"),
    ("Ed", "Garth"),
    ("Fernando", "Garth"),
    ("Fernando", "Heather"),
    ("Garth", "Heather"),
    ("Heather", "Ike"),
    ("Ike", "Luis")
]

edges
```

## Graph


```python
G = nx.Graph()
G.add_nodes_from(nodes)
G.add_edges_from(edges)

print(f"Number of nodes: {G.number_of_nodes()}")
print(f"Number of edges: {G.number_of_edges()}")
```

```python
#positions sourced
#For purpose of this assignment I have recreated the Krackhardt Kite but using this ring to branch off other nodes resembles chemistry
pos = {
    "Carol": (1, 3),
    "Andre": (0, 2),
    "Fernando": (2, 2),
    "Diane": (1, 1.4),
    "Beverly": (0, 0.5),
    "Garth": (2, 0.5),
    "Ed": (1, -0.5),
    "Heather": (3, 1.2),
    "Ike": (4, 1.2),
    "Luis": (5, 1.2)
}

plt.figure(figsize=(10, 6))

nx.draw(
    G,
    pos=pos,
    with_labels=True,
    node_size=1800,
    font_size=11,
    font_weight="bold",
    width=2
)

plt.title("Krackhardt Kite Graph")
plt.show()
```
